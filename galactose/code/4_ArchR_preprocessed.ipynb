```r
ArrowFiles <- createArrowFiles(
  inputFiles = paste0("../data/rep", 1:2, "/cellranger/fragments.tsv.gz"),
  sampleNames = paste0("rep", 1:2),
  minTSS = 2,
  minFrags = 1000, 
  addTileMat = TRUE,
  addGeneScoreMat = TRUE
)
```

```r
Gal <- ArchRProject(
  ArrowFiles = ArrowFiles, 
  outputDirectory = "../output/POLG_GAL",
  copyArrows = TRUE
)
```
